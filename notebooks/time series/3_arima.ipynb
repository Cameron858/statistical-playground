```python
import numpy as np
from statsmodels.tsa.arima_process import ArmaProcess
import matplotlib.pyplot as plt

np.random.seed(19)
```

# ARIMA: AutoRegressive Integrated Moving Average

## AutoRegressive (AR)

An $ AR(p) $ model expresses the current value $ y_t $ as a linear combination of its past $ p $ values plus noise $ \epsilon $:
$$ y_t = c + \sum_{i=1}^{p}\phi_ty_{t-i} + \epsilon_t $$
Where:
- $ p $ = AR order
- $ c $ = Constant
- $ \phi_i $ = AR coefficients
- $ \epsilon_t $ ~ $ \mathcal{N}(0, \sigma^2) $ is white noise

Below, we show an AR(1) process and an AR(2) process.

```python
n = 100

# statsmodels.tsa.arima_process.ArmaProcess expects both AR & MA params,
# including the zero-lag coefficient (always 1).
# For AR(1) with phi = 0.75, we use [1, -0.75] (note the negative sign).
ar1_params = [1, -0.75]  # AR polynomial: 1 - 0.75*L
ar1 = ArmaProcess(ar1_params, [1])
ar1_values = ar1.generate_sample(n)

ar2_params = [1, -0.75, -0.23]  # AR polynomial: 1 - 0.75*L
ar2 = ArmaProcess(ar2_params, [1])
ar2_values = ar2.generate_sample(n)

fig, axes = plt.subplots(2, 1, figsize=(8, 5), sharex=True)
axes[0].plot(ar1_values, label=r"$ y_t = 0.75 y_{t-1} + \epsilon_t $")
axes[0].set_title("Simulated AR(1) Process")
axes[0].legend()

axes[1].plot(ar2_values, label=r"$ y_t = 0.75 y_{t-1} + 0.23 y_{t-2} + \epsilon_t $")
axes[1].set_title("Simulated AR(2) Process")
axes[1].legend()

fig.tight_layout()
```

## Moving Average (MA)

```python
n = 100

# For MA(q), AR params = [1] (no AR component), MA params use positive sign for theta terms.
ma1_params = [1, 0.65]  # MA polynomial: 1 + 0.75*L
ma1 = ArmaProcess([1], ma1_params)
ma1_values = ma1.generate_sample(n)

ma2_params = [1, 0.65, -0.4]  # MA polynomial: 1 + 0.75*L + 0.23*L^2
ma2 = ArmaProcess([1], ma2_params)
ma2_values = ma2.generate_sample(n)

fig, axes = plt.subplots(2, 1, figsize=(8, 5), sharex=True)
axes[0].plot(ma1_values, label=r"$ y_t = \epsilon_t + 0.65 \epsilon_{t-1} $")
axes[0].set_title("Simulated MA(1) Process")
axes[0].legend()

axes[1].plot(ma2_values, label=r"$ y_t = \epsilon_t + 0.65 \epsilon_{t-1} - 0.4 \epsilon_{t-2} $")
axes[1].set_title("Simulated MA(2) Process")
axes[1].legend()

fig.tight_layout()
```
